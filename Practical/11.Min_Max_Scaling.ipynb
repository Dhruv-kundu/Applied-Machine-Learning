```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("loan.csv")
dataset.head(50)
```

```python
# Feature Scaling --> Normalization --> Min-Max Scalar
"""
It is a scaling technique in which values are shifted and rescaled so that they end up ranging between 0 and 1 . 
X new = X i -min(x)/ max(x) - min(x)
"""
```

```python
from sklearn.preprocessing import MinMaxScaler
ms = MinMaxScaler()
ms.fit(dataset[["CoapplicantIncome"]]) #2d 
dataset["CoapplicantIncome_ss"] = ms.transform(dataset[["CoapplicantIncome"]])
```

```python
dataset
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title("Before")
sns.distplot(dataset["CoapplicantIncome"])
plt.subplot(1,2,2)
plt.title("After")
sns.distplot(dataset["CoapplicantIncome_ss"])
plt.show()
```
