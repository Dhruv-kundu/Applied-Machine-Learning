```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

import numpy as np
```

```python
dataset = pd.read_csv("loan.csv")
dataset.head(10)
```

```python
# Function Transformer
# It is a tool used to apply a custom function (transformation) to data in a machine learning pipeline.

# Dataset -> nan value remove --> outlier remove ---> work  ( upcoming -- train_test_spliting)
# Ideal work flow
```

```python
from sklearn.preprocessing import FunctionTransformer
ft = FunctionTransformer(func= np.log1p) # lamba x:x**2
ft.fit(dataset[["CoapplicantIncome"]]) #2d
dataset[["CoapplicantIncome_ft"]] = ft.transform(dataset[["CoapplicantIncome"]])
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title("Before")
sns.distplot(dataset["CoapplicantIncome"])
plt.subplot(1,2,2)
plt.title("After")
sns.distplot(dataset["CoapplicantIncome_ft"])
plt.show()
```

```python
#| Transformer            | Purpose                                          | Example                     |
#| ---------------------- | ------------------------------------------------ | --------------------------- |
#| `StandardScaler()`     | Standardizes data (mean = 0, std = 1)            | Heights, weights            |
#| `MinMaxScaler()`       | Scales values to a fixed range (usually 0–1)     | Pixel values                |
#| `LabelEncoder()`       | Converts labels to integers                      | Male/Female → 0/1           |
#| `OneHotEncoder()`      | Converts categorical variables to binary columns | Red → (1,0,0)               |
#| `OrdinalEncoder()`     | Converts ordered categories to numbers           | Low, Medium, High → 0,1,2   |
```
