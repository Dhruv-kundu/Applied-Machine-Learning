```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv("placement.csv")
dataset.head(3)
```

```python
dataset.isnull().sum()
```

```python
x = dataset[["cgpa"]]
x.ndim  # int object is not call able therefore just .ndim not .ndim()
```

```python
y = dataset["package"]
y.ndim
```

```python
sns.scatterplot(x="cgpa",y="package", data=dataset)  #linear
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test ,  y_train , y_test = train_test_split(x, y ,random_state=42 , test_size=0.21)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train , y_train)
```

```python
lr.predict([[6.89]]) #3.26 actual
```

```python
lr.score(x_test , y_test) * 100
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
sns.scatterplot(x="cgpa",y="package", data=dataset)
plt.plot(dataset["cgpa"] , lr.predict(x),c ="red")
plt.legend(["org data","predict data"])
#plt.savefig("predict.jpg")
plt.show()
```
