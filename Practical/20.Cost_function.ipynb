```python
"""
What is a Cost Function?
A Cost Function (or Loss Function) is a mathematical function that measures how far the model's predictions are from the actual values.

In simple words:
A cost function tells us how much error the model is making.

The goal of machine learning is to minimize the cost function so that predictions become as accurate as possible.

Types 
1. Regression Cost Function      2. Classification Cost Function

Regression Concept
In regression, the output is a continuous value.
Examples:
- House Price Prediction
- Salary Prediction
- Temperature Prediction
Mean Squared Error (MSE) - Most Common
    Used in Linear Regression
    Squares errors, so large errors are penalized more
Root Mean Squared Error (RMSE)
    Same units as target variable
Mean Absolute Error (MAE)
    Less sensitive to outliers

Binary Classification Cost Function:
    Measures the error between actual and predicted probabilities for two classes using Binary Cross Entropy.
Multi-Class Classification Cost Function:
    Measures the error between actual and predicted probabilities for multiple classes using Categorical Cross Entropy.
"""
```
