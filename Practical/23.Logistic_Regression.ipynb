```python
"""
Logistic Regression
✅ Logistic Regression is a supervised learning classification algorithm used to predict categorical outcomes based on input features.
✅ It is used for predicting a categorical dependent variable using a given set of independent variables (features).
✅ Therefore, the output (dependent variable) must be categorical or discrete in nature. It can be Yes/No, 0/1, True/False, Pass/Fail, etc.
✅ Instead of directly predicting the exact value 0 or 1, Logistic Regression predicts the probability of an event occurring, which lies between 0 and 1. Based on a threshold value (usually 0.5), the final class label is assigned.
✅ Logistic Regression assumes a linear relationship between the input features and the log-odds of the target variable.

Types of Logistic Regression
1. Binomial Logistic Regression
There are only two possible categories of the dependent variable.
Examples:
0 or 1
Pass or Fail
Yes or No
True or False

2. Multinomial Logistic Regression
There are three or more possible unordered categories of the dependent variable.
Examples:
Cat, Dog, Sheep
Red, Blue, Green
Car, Bike, Bus
Keyword: Unordered Categories

3. Ordinal Logistic Regression
There are three or more possible ordered categories of the dependent variable.
Examples:
Low, Medium, High
Poor, Average, Good, Excellent
Small, Medium, Large
Keyword: Ordered Categories
"""
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np 
```

```python
dataset = pd.read_csv("Social_Network_Ads.csv")
dataset.head(3)
```

```python
dataset.isnull().sum()
```

```python
dataset.drop(columns=["EstimatedSalary"],inplace=True)
sns.scatterplot(x="Age",y="Purchased", data=dataset)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x= dataset[["Age"]]
y = dataset["Purchased"]
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(x_train , y_train)
lr.score(x_test,y_test)*100
```

```python
sns.scatterplot(x="Age",y="Purchased" , data= dataset)
sns.lineplot(x="Age" , y=lr.predict(x),data=dataset , color= "red")
plt.show()
```
