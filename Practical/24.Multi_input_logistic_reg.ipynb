```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np 
```

```python
dataset = pd.read_csv("placement (1).csv")
dataset.head(3)
```

```python
dataset.isnull().sum()
```

```python
#dataset.drop(columns=["Student_ID"],inplace=True)
sns.scatterplot(x="CGPA",y="IQ", data=dataset, hue="Placement")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x= dataset.iloc[:,:-1]
y = dataset["Placement"]
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(x_train , y_train)
lr.score(x_test,y_test)*100
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(x.to_numpy(),y.to_numpy(),clf=lr)
plt.show()
"""
clf stands for Classifier.
It is the trained machine learning model.
Here lr is a trained Logistic Regression model.

What is a Decision Boundary?
A decision boundary is a line (or hyperplane in higher dimensions) that separates different classes predicted by a classification model. In Logistic Regression,
the decision boundary is linear and is formed where the predicted probability equals the threshold value (typically 0.5).
"""
```
