```python
# Confusion Matrix
# Definition
# - A Confusion Matrix is a simple and useful tool used to evaluate the performance of a classification model in Machine Learning.
# - It helps us understand how well a model is classifying data into different categories.
# - It is also known as an Error Matrix.
# - The matrix summarizes the model's prediction results by showing:
#      Correct Predictions
#      Incorrect Predictions
# - It is commonly used for Binary Classification problems (0 and 1 classes).
```

```python
# 
# How Confusion Matrix is Generated
# 1. Collect the dataset.
# 2. Split the dataset into:
#     Training Data
#     Testing Data
# 3. Train the classification model using the training data.
# 4. Use the trained model to make predictions on the test data.
# 5. Compare:
#     Actual Values (y_test)
#     Predicted Values (y_pred)
# 6. Generate the Confusion Matrix to analyze the model's performance.
```

```python
# workflow
# Dataset
#    ↓
# Train-Test Split
#    ↓
# Model Training
#    ↓
# Prediction on Test Data
#    ↓
# Compare Actual vs Predicted
#    ↓
# Confusion Matrix
```

```python
"""
Terms
TP (True Positive): Predicted Positive and actually Positive.
TN (True Negative): Predicted Negative and actually Negative.
FP (False Positive): Predicted Positive but actually Negative.
FN (False Negative): Predicted Negative but actually Positive.
"""
```

```python
# Accuracy
"""
Measures the overall correctness of the model.
Formula
Accuracy=TP+TN / TP+TN+FP+FN
"""
# Error
"""
Formula
Accuracy=FP+FN / TP+TN+FP+FN
"""
# Precision
"""
Measures how many predicted positives are actually positive.
Formula
Precision= TP/TP+FP
"""
# Recall (Sensitivity or True Positive Rate)
"""
Measures how many actual positives were correctly identified.
Formula
Recall= TP/TP+FN
"""
# Precision (rise) FP(Decrease)
# Recall (rise) Fn(Decrease)

# F1 Score
"""
Harmonic mean of Precision and Recall.
Formula
F1= 2 *( Precision * Recall / Precision+Recall)
"""
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np 
```

```python
dataset = pd.read_csv("placement (1).csv")
dataset.head(3)
```

```python
from sklearn.model_selection import train_test_split
x= dataset.iloc[:,:-1]
y = dataset["Placement"]
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(x_train , y_train)
lr.score(x_test,y_test)*100
```

```python
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score
cf = confusion_matrix(y_test , lr.predict(x_test))
cf
```

```python
sns.heatmap(cf,annot=True)
plt.show()
# annot stands for annotation.
# It displays the numerical values inside each cell of the heatmap.
```

```python
precision_score(y_test , lr.predict(x_test))*100 , recall_score(y_test , lr.predict(x_test))*100 , f1_score(y_test , lr.predict(x_test))*100
```

```python
"""
| Score       | Interpretation    |
| ----------- | ----------------- |
| 1.0 (100%)  | Perfect           |
| 0.90 - 0.99 | Excellent         |
| 0.80 - 0.89 | Good              |
| 0.70 - 0.79 | Acceptable        |
| Below 0.70  | Needs improvement |
"""
```

```python
from sklearn.metrics import accuracy_score
print("Accuracy :", accuracy_score(y_test, lr.predict(x_test)) * 100)
print("Error Rate :", (1 - accuracy_score(y_test, lr.predict(x_test))) * 100)
```
