```python
# Naive Bayes Algorithm
"""
What is Naive Bayes?
Naive Bayes is a supervised machine learning classification algorithm based on Bayes' Theorem. It is widely used for classification tasks such as spam detection, sentiment analysis, and document categorization.

Why is it called "Naive Bayes"?
    Naive: It assumes that all input features are independent of each other.
    Bayes: It is based on Bayes' Theorem, which calculates the probability of an event using prior knowledge.

Probability
Probability measures the likelihood of an event occurring.
Formula : P(E)= Favorable Outcomes/ Total Outcomes

Independent Probability
Two events are independent if the occurrence of one event does not affect the occurrence of the other.
Example: Rolling a Dice
Each number has an equal probability:
P(1)= 1/6
The probability of getting a specific number remains the same regardless of previous rolls.

Dependent Probability
Two events are dependent if the occurrence of one event affects the probability of the other.
Example: Balls in a Box
Suppose a box contains:
3 Red Balls
2 Black Balls
Total balls = 5
P(Red) = 3/5 P(Blue)=2/5
Now assume one red ball is drawn and not replaced.
P(Red) = 2/4 P(Blue)=2/4
Conditional Probability

Conditional probability is the probability of an event occurring given that another event has already occurred.

Formula
P(A∣B)= P(A∩B) / P(B)
P(A∣B) = Probability of A given B
P(A∩B) = Probability of both A and B occurring
P(B) = Probability of event B

Suppose:
A = Student passes the exam
B = Student studied regularly
Then:
P(A∣B)
represents the probability that a student passes the exam given that they studied regularly.

Bayes' Theorem
Bayes' Theorem is based on conditional probability. It helps determine the probability of a hypothesis using prior knowledge.
Formula
P(A∣B)= P(B∣A)×P(A) / P(B)

Where:
P(A∣B) = Posterior Probability
P(B∣A) = Likelihood
P(A) = Prior Probability
P(B) = Evidence

Bayes' Theorem Example (A and B)
Let:
A = Person has a disease
B = Test result is positive
Then
P(A∣B)= P(B∣A)×P(A) / P(B)
This calculates the probability that a person actually has the disease given that the test result is positive.
"""
```

```python
"""
Types of Naive Bayes Models
There are three main types of Naive Bayes models:
1. Gaussian Naive Bayes
Used for continuous numerical data.
Assumes data follows a normal (Gaussian) distribution.
Commonly used for datasets containing measurements such as height, weight, temperature, etc.
Example: Age, Salary, Height

2. Multinomial Naive Bayes (boolean)
Used for discrete count data.
Commonly used in text classification.
Works on word frequencies or counts.
Example: Spam detection, Sentiment analysis,Document classification

3. Bernoulli Naive Bayes
Used for binary features.
Features can only have two values: 0 or 1, Yes or No, True or False.
"""
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np 
from mlxtend.plotting import plot_decision_regions
```

```python
dataset = pd.read_csv("Placement2.csv")
dataset.head(3)
```

```python
sns.scatterplot(x="cgpa",y="score",data=dataset,hue="placed")
plt.show()
```

```python
x = dataset.iloc[:,:-1]
y = dataset["placed"]
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y, test_size=0.2 , random_state=42)
sns.kdeplot(data=dataset["score"])
plt.show()
```

```python
sns.kdeplot(data=dataset["cgpa"])
plt.show()
```

```python
from sklearn.naive_bayes import GaussianNB , MultinomialNB , BernoulliNB
gnb = GaussianNB()
gnb.fit(x_train,y_train)
gnb.score(x_test , y_test)*100 , gnb.score(x_train,y_train)*100
```

```python
plot_decision_regions(x.to_numpy(),y.to_numpy(), clf=gnb)
plt.show()
```

```python
mnb = MultinomialNB()
mnb.fit(x_train,y_train)
mnb.score(x_test , y_test)*100 , mnb.score(x_train,y_train)*100
```

```python
plot_decision_regions(x.to_numpy(),y.to_numpy(), clf=mnb)
plt.show()
```

```python
bnb = BernoulliNB()
bnb.fit(x_train,y_train)
bnb.score(x_test , y_test)*100 , bnb.score(x_train,y_train)*100
```

```python
plot_decision_regions(x.to_numpy(),y.to_numpy(), clf=bnb)
plt.show()
```
