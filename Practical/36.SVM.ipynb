```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
# SVM (Support Vector Machine)
"""
Support Vector Machine (SVM) is a supervised machine learning algorithm used for:
1. Classification
2. Regression (called Support Vector Regression - SVR)
"SVM is primarily used for classification problems, but it can also be extended to regression problems through Support Vector Regression (SVR)."

Hard Margin vs Soft Margin
- Hard Margin SVM
    Assumes data is perfectly linearly separable.
    No misclassification is allowed.
    Sensitive to outliers.
- Soft Margin SVM
    Allows some misclassification.
    More robust to noise and outliers.
    Better generalization on unseen data.

SVM Maximum Margin Diagram
                    X2
                     ↑

      Green ●                 ● Green
                    + HYPERPLANE
             -------------------------
                    Margin (+)

      Green ●     *Support Vector*

             -------------------------
                  Decision Boundary
                    (Hyperplane)

      Blue ●      *Support Vector*

             -------------------------
                    Margin (-)

      Blue ●                 ● Blue

------------------------------------------------→ X1

Detailed SVM Diagram
                     X2
                      ↑

     Green ●     Green ●      Green ●

     Positive Hyperplane
     y = mx + c + δ
     -------------------------

            ↑
            │ d
            │

     Decision Boundary
     y = mx + c
     -------------------------

            │
            │ d
            ↓

     Negative Hyperplane
     y = mx + c - δ
     -------------------------

     Blue ●      Blue ●      Blue ●

------------------------------------------------→ X1

Where:
d = Distance from hyperplane to margin
Margin Width = 2d
SVM tries to maximize:
Maximum Margin = 2d

Support Vectors
Support vectors are the data points closest to the hyperplane.

"""
```

```python
"""
Maximum Margin vs Minimum Margin
Maximum Margin (Good)
Green ●         ●

------------------------
       Large Margin
------------------------

Blue ●          ●


Minimum Margin (Bad)
Green ●    ●

-------------
 Small Margin
-------------

Blue ●     ●
"""
# Maximum margin generally gives better generalization.
```

```python
dataset = pd.read_csv("placement2.csv")
dataset.head(3)
```

```python
dataset.isnull().sum()
```

```python
sns.scatterplot(x="cgpa",y="score",data=dataset,hue="placed")
plt.show()
```

```python
x= dataset.iloc[:,:-1]
y= dataset["placed"]
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.svm import SVC
sc = SVC()
# kernel : {'linear', 'poly', 'rbf', 'sigmoid', 'precomputed'} or callable,          default='rbf'
#     Specifies the kernel type to be used in the algorithm. If
#     none is given, 'rbf' will be used. If a callable is given it is used to
#     pre-compute the kernel matrix from data matrices; that matrix should be
#     an array of shape ``(n_samples, n_samples)``. For an intuitive
#     visualization of different kernel types see
sc.fit(x_train,y_train)
sc.score(x_test,y_test)*100
```

```python
#Overfitting
sc.score(x_test,y_test)*100,sc.score(x_train,y_train)*100,sc.score(x_test,y_test)*100-sc.score(x_train,y_train)*100
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(x.to_numpy(),y.to_numpy(), clf=sc)
plt.show()
```
