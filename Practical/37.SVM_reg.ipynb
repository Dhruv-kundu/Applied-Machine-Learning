```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
dataset = pd.read_csv("placement2.csv")
dataset.head(3)
```

```python
x= dataset.iloc[:,:-1]
y= dataset["placed"]
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.svm import SVR
sc = SVR()
# kernel : {'linear', 'poly', 'rbf', 'sigmoid', 'precomputed'} or callable,          default='rbf'
#     Specifies the kernel type to be used in the algorithm. If
#     none is given, 'rbf' will be used. If a callable is given it is used to
#     pre-compute the kernel matrix from data matrices; that matrix should be
#     an array of shape ``(n_samples, n_samples)``. For an intuitive
#     visualization of different kernel types see
sc.fit(x_train,y_train)
sc.score(x_test,y_test)*100
```

```python
#Overfitting
sc.score(x_test,y_test)*100,sc.score(x_train,y_train)*100,sc.score(x_test,y_test)*100-sc.score(x_train,y_train)*100
```
