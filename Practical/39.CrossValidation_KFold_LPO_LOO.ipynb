```python
# Cross Validation in Machine Learning
"""
Cross Validation is the technique for validating model efficiency by training it on a subset of data and testing it on previously unseen data.
Why Cross Validation?
- Reduces overfitting
- Provides better performance estimation
- Makes efficient use of data
- Helps in model selection and hyperparameter tuning

Workflow
                    DATASET
                       │
                       ▼
             Train-Test Split
                       │
         ┌─────────────┴─────────────┐
         │                           │
         ▼                           ▼
    Training Data              Testing Data
         │
         ▼
   Cross Validation
         │
         ▼
 Hyperparameter Tuning
         │
         ▼
  Best Parameters Found
         │
         ▼
 Retrain Model on Full
   Training Dataset
         │
         ▼
      Final Model
         │
         ▼
 Evaluate on Test Data
         │
         ▼
   Final Performance

Methods of Cross Validation
1. Hold-Out Cross Validation
2. K-Fold Cross Validation
3. Stratified K-Fold Validation
4. Leave-One-Out Cross Validation (LOOCV)
5. Leave-P-Out Cross Validation (LPOCV)
6. Repeated Randomized Subsampling Validation
7. Time Series Cross Validation
8. Nested Cross Validation
"""
```

```python
# 1. Hold-Out Cross Validation
"""
The dataset is split once into:
Training Data
Validation Data
Testing Data

Diagram
Dataset
   │
   ▼

┌────────┬─────────┬────────┐
│ Train  │ Valid   │ Test   │
│ 70%    │ 15%     │ 15%    │
└────────┴─────────┴────────┘
"""
```

```python
# 2. K-Fold Cross Validation
"""
The original dataset is partitioned into K folds.
In each iteration, one fold is used for validation and K−1 folds are used for training.

Example: K = 5
Fold1 Fold2 Fold3 Fold4 Fold5
Iteration 1
[V]   [T]   [T]   [T]   [T]
Iteration 2
[T]   [V]   [T]   [T]   [T]
Iteration 3
[T]   [T]   [V]   [T]   [T]
Iteration 4
[T]   [T]   [T]   [V]   [T]
Iteration 5
[T]   [T]   [T]   [T]   [V]

Final Score:
Average of all 5 validation scores

Disadvantage
❌ Not suitable for highly imbalanced datasets.
"""
```

```python
# 3. Stratified K-Fold Cross Validation
"""
Definition : A variation of K-Fold where the proportion of each class is maintained in every fold.
Why? : Solves the imbalance problem.
Example
Original Dataset
Class A = 90%
Class B = 10%

Stratified Folds
Fold 1 → 90% A, 10% B
Fold 2 → 90% A, 10% B
Fold 3 → 90% A, 10% B
Fold 4 → 90% A, 10% B
Fold 5 → 90% A, 10% B

Diagram
Dataset
AAAAAAAAB

       │

       ▼

Fold1  AAAAB
Fold2  AAAAB
Fold3  AAAAB
Fold4  AAAAB
Fold5  AAAAB

Advantage
✅ Best choice for classification problems with imbalanced classes.
"""
```

```python
# Leave-One-Out Cross Validation (LOOCV)
"""
A special case of Leave-P-Out Cross Validation where: P = 1
✅ LOOCV is an exhaustive cross-validation technique.
Working
Suppose: Dataset Size = 5
Iteration 1
Train: D2 D3 D4 D5
Test : D1
Iteration 2
Train: D1 D3 D4 D5
Test : D2
Iteration 3
Train: D1 D2 D4 D5
Test : D3
and so on...
Diagram
[D1 D2 D3 D4 D5]
Iteration 1
 Test = D1
Iteration 2
 Test = D2
Iteration 3
 Test = D3
Iteration 4
 Test = D4
Iteration 5
 Test = D5

Advantages
Uses maximum training data.
Low bias.
High accuracy estimate.

Disadvantages
Very slow.
Computationally expensive.
"""
```

```python
# 5. Leave-P-Out Cross Validation
"""
An exhaustive cross-validation technique where P observations are used as validation data.

Example
Dataset Size = 10
P = 2
Diagram
D1 D2 D3 D4 D5 D6 D7 D8 D9 D10
Iteration 1
Test : D1 D2
Iteration 2
Test : D1 D3
Iteration 3
Test : D1 D4
...
Every possible combination is tested.
Advantages
Thorough evaluation

Disadvantages
Extremely expensive for large datasets
"""
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
dataset = pd.read_csv("placement2.csv")
dataset.head(3)
```

```python
x= dataset.iloc[:,:-1]
y= dataset["placed"]
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.model_selection import LeaveOneOut , LeavePOut , KFold , StratifiedKFold
lo = LeaveOneOut()
for train, test in lo.split(x, y):
    print("Train:", train)
    print("Test :", test)
    print()
```

```python
lp = LeavePOut(p=2)
for train, test in lp.split(x, y):
    print("Train:", train)
    print("Test :", test)
    print()
```

```python
kf = KFold(n_splits=5)
for train, test in kf.split(x, y):
    print("Train:", train)
    print("Test :", test)
    print()
```

```python
skf = StratifiedKFold(n_splits=5)
for train, test in skf.split(x, y):
    print("Train:", train)
    print("Test :", test)
    print()
```

```python
from sklearn.model_selection import cross_val_score, KFold
from sklearn.linear_model import LinearRegression

kf = KFold(n_splits=10,shuffle=True,random_state=42)
p = cross_val_score(LinearRegression(),x,y,cv=kf,scoring='r2')
print(p)
```
