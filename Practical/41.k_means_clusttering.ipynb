```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np 
import seaborn as sns
```

```python
"""
K-Means Clustering
K-Means is an unsupervised machine learning algorithm used to group unlabeled data into K clusters based on similarity.
K = Number of clusters to be formed.
The algorithm tries to minimize the distance between data points and their cluster centroids.

Steps of K-Means
1. Randomly select K centroids (center points).
2. Assign each data point to the nearest centroid (Grouping).
3. Calculate the mean of each cluster and move the centroid to the new mean position.
4. Repeat Steps 2 and 3 until the centroids stop changing significantly, resulting in perfect/stable clusters.

WCSS (Within-Cluster Sum of Squares)
WCSS measures the total variation within clusters.
WCSS=i∈C1​∑​(Pi​−C1​)2+i∈C2​∑​(Pi​−C2​)2+⋯
Where:
Pi= Data point
Ci= Cluster centroid
"""
# Lower WCSS indicates tighter and better clusters.
```

```python
# Elbow Method
"""
The Elbow Method helps determine the optimal value of K.

Train K-Means for different values of K.
Calculate WCSS for each K.
Plot K vs WCSS.
Choose the K at the elbow point, where the decrease in WCSS becomes slower.
Elbow Diagram
WCSS
 ^
 |
 |\
 | \
 |  \
 |   \
 |    \
 |     \__
 |        \__
 |           \__
 +--------------------> Number of Clusters (K)
       ^
       |
     Elbow
   (Optimal K)

WCSS
1000 |*
 900 |
 800 | *
 700 |
 600 |  *
 500 |
 400 |    *
 300 |      *
 200 |        *
 100 |          *
     +----------------
      1 2 3 4 5 6 7 8
            ^
            |
       Optimal K = 4
"""
```

```python
# K-Means++
"""
K-Means++ is an improved version of K-Means.
Problem with standard K-Means:
Random centroid selection can lead to poor clustering and slower convergence.

Solution:
K-Means++ chooses initial centroids more intelligently, spreading them far apart.

Benefits:
Better cluster quality.
Faster convergence.
Reduces chances of getting stuck in a poor solution.
"""
# Important Parameters

# n_clusters → Number of clusters (K)
# init='k-means++' → Uses K-Means++ initialization
# inertia_ → Returns WCSS value
```

```python
dataset = pd.read_csv("iris.raw.csv")
dataset.head(3)
```

```python
sns.pairplot(data=dataset)
plt.show()
```

```python
from sklearn.cluster import KMeans
wcss=[]
for i in range(2,21):
    km = KMeans(n_clusters=i,init='k-means++')
    km.fit(dataset)
    wcss.append(km.inertia_) #values of wcss
    
```

```python
wcss
```

```python
plt.figure(figsize=(10,5))

plt.plot(range(2,21), wcss, marker='o')
plt.xlabel("No. of Clusters")
plt.ylabel("WCSS")
plt.xticks(range(2,21))
plt.grid(axis='x')

plt.show()
```

```python
kmn = KMeans(n_clusters=3)
dataset["Predict"]= kmn.fit_predict(dataset)
```

```python
kmn1 = KMeans(n_clusters=4)
dataset["Predict1"]= kmn1.fit_predict(dataset)
```

```python
kmn2 = KMeans(n_clusters=5)
dataset["Predict2"]= kmn2.fit_predict(dataset)
```
