```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
import numpy as np
```

```python
dataset = pd.read_csv("iris.raw.csv")
dataset.head(3)
```

```python
sns.pairplot(data=dataset)
plt.show()
```

```python
from sklearn.cluster import KMeans
wcss=[]
for i in range(2,21):
    km = KMeans(n_clusters=i,init='k-means++')
    km.fit(dataset)
    wcss.append(km.inertia_) #values of wcss
    
```

```python
from sklearn.metrics import silhouette_score
# silhouette_score(dataset, labels=km.labels_)
```

```python
ss = []
no_c = [j for j in range(2,21)]
for i in range(2,21):
    km1 = KMeans(n_clusters=i)
    km1.fit(dataset)
    ss.append(silhouette_score(dataset, labels=km1.labels_))
```

```python
plt.plot(no_c,ss)
plt.xlabel("ss")
plt.ylabel("no. of clusters")
plt.xticks(no_c)
plt.grid(axis='x')
plt.show()
```
