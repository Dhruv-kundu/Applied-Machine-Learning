```python
# Ensemble Learning
"""
Ensemble Learning is a machine learning technique that combines the insights or predictions obtained from multiple learning models to facilitate more accurate and improved decisions.
There are two main types of ensemble learning:
1. Bagging – Bootstrap Aggregation
2. Boosting

Basic Ensemble Techniques
The basic ensemble techniques are:
- Max Voting
- Averaging
- Weighted Averaging

The advanced ensemble techniques are:
- Stacking
- Blending
- Bagging
- Boosting

Algorithms Based on Bagging
Some algorithms based on bagging are:
- Bagging Meta-Estimator
- Random Forest
- Algorithms Based on Boosting

Some popular boosting algorithms are:
- AdaBoost
- GBM (Gradient Boosting Machine)
- XGBoost
- LightGBM
- CatBoost

Max Voting
The Max Voting method is generally used for classification problems.
In this technique, multiple models are used to make a prediction for each data point. The prediction made by each model is considered as a vote.
The prediction that receives the majority of the votes from the models is taken as the final prediction.

For example, suppose we have three models:
Decision Tree → Class A
SVM → Class B
KNN → Class A

Here:
Class A = 2 votes
Class B = 1 vote

Therefore, the final prediction will be:
Class A

Example
Suppose we have a dataset of 10,000 rows, where:
8,000 rows are used for training.
2,000 rows are used for testing.

For training, we use three models:
Decision Tree
SVM
KNN
Each model makes its own prediction, and the predictions can then be combined using an ensemble technique such as voting or averaging.

Averaging
Averaging means taking the average of the predictions made by all the models and using it to make the final prediction.
Averaging is generally used for regression problems, where the output is a numerical value. It can also be used for classification when we are averaging the predicted probabilities.
For example, suppose three models give the following predictions:
Decision Tree = 10
SVM = 12
KNN = 11
Then: Final Prediction= 10+12+11/3
Final Prediction=11

Therefore, the final prediction obtained by averaging is 11.
In classification, instead of directly averaging class labels, we can average the predicted probabilities from the different models and use the class with the highest probability as the final prediction."""
```
