```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
dataset = pd.read_csv("placement.csv")
dataset.head(3)
```

```python
x = dataset.iloc[:,:-1]
y = dataset["package"]
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y, random_state=42 , test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
```

```python
lr = LinearRegression()
lr.fit(x_train , y_train)
lr.score(x_test , y_test)*100 , lr.score(x_train , y_train)*100 , lr.score(x_test , y_test)*100-lr.score(x_train , y_train)*100
```

```python
dt = DecisionTreeRegressor()
dt.fit(x_train , y_train)
dt.score(x_test , y_test)*100 , dt.score(x_train , y_train)*100 , dt.score(x_test , y_test)*100-dt.score(x_train , y_train)*100
```

```python
sv = SVR()
sv.fit(x_train , y_train)
sv.score(x_test , y_test)*100 , sv.score(x_train , y_train)*100 , sv.score(x_test , y_test)*100-sv.score(x_train , y_train)*100
```

```python
from sklearn.ensemble import VotingRegressor
li = [("dt1",DecisionTreeRegressor()),("sv1",SVR()),("lr",LinearRegression())]
vr = VotingRegressor(li) #weights = [10,4,7]
vr.fit(x_train , y_train)
vr.score(x_test , y_test)*100 , vr.score(x_train , y_train)*100 , vr.score(x_test , y_test)*100-vr.score(x_train , y_train)*100
```

```python
prd ={"dt":dt.predict(x_test),"svc":sv.predict(x_test),"lr": lr.predict(x_test),"vr":vr.predict(x_test)}
pd.DataFrame(prd)
```
