```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
dataset = pd.read_csv("placement.csv")
dataset.head(3)
```

```python
x = dataset.iloc[:,:-1]
y = dataset["package"]
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y, random_state=42 , test_size=0.2)
```

```python
from sklearn.ensemble import BaggingRegressor , RandomForestRegressor
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(x_train , y_train)
lr.score(x_test , y_test)*100 , lr.score(x_train , y_train)*100 , lr.score(x_test , y_test)*100-lr.score(x_train , y_train)*100
```

```python
bg = BaggingRegressor(estimator=LinearRegression(),n_estimators=50)
bg.fit(x_train , y_train)
bg.score(x_test , y_test)*100 , bg.score(x_train , y_train)*100 , bg.score(x_test , y_test)*100-bg.score(x_train , y_train)*100
```

```python
rf = RandomForestRegressor(n_estimators=50)
rf.fit(x_train , y_train)
rf.score(x_test , y_test)*100 , rf.score(x_train , y_train)*100 , rf.score(x_test , y_test)*100-rf.score(x_train , y_train)*100
```
