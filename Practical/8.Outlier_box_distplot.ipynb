```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("loan.csv")
dataset.head(10)
```

```python
# Box plot |--   box   --|
# min_range     Q1 Q3     max_range
# Q1 = 25 quantile q3 = 75 quantile
# IQR == inter quantile range = Q3- Q1
# min = Q1 - (1.5 * IQR)         max = Q1 + (1.5 * IQR)
```

```python
dataset.describe()
```

```python
dataset.info()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(x="CoapplicantIncome",data=dataset)
plt.show()
```

```python
sns.distplot(dataset["CoapplicantIncome"])
plt.show()
```

```python
dataset.shape
```

```python
q1 = dataset["CoapplicantIncome"].quantile(0.25)
q3 = dataset["CoapplicantIncome"].quantile(0.75)

q1 , q3
```

```python
IQR = q3-q1
min_range = q1 -(1.5 * IQR)
max_range = q1 +(1.5 * IQR)

min_range , max_range
```

```python
new = dataset[dataset["CoapplicantIncome"]<=max_range]
new.shape
```

```python
sns.boxplot(x="CoapplicantIncome", data=new)
plt.show()
```

```python
sns.distplot(new["CoapplicantIncome"])
plt.show()
```
